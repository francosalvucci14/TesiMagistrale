```python
import temporal_forest

print("Creazione del grafo...")
# Crei l'oggetto C++ direttamente da Python!
G = temporal_forest.TemporalGraph()

# Puoi caricare il dataset leggendo il file in Python 
# e inserendo gli archi comodamente
with open("chess_year", "r") as f:
    for line in f:
        u, v, t = line.strip().split()
        G.add_edge(u, v, int(t))

print("Costruzione della foresta (velocità C++)...")
# TRF costruirà tutti gli alberi binari sfruttando il tuo codice nativo
TRF = temporal_forest.TemporalRangeForest(G)

print("Esecuzione della query...")
# Esegui la query per il nodo 862 nell'intervallo 1998-2002
sketch_risultante = TRF.query("862", 1998, 2002)

#print(f"Risultato della query: {sketch_risultante.jaccard_sim()}")
```
